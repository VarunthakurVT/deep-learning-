```python
input=[[1.3,3.4,4.4],
       [4.3,4.4,6.3],
       [4.2,4.3,6.3]]

weight=[[2.3,3.5,2.4],  
        [3.3,6.5,3.4],  
        [5.3,4.5,5.3]]  

bias=[2,3,4]   
outputs=[]  
# Process the first input sample
sample = input[0]
layer_outputs=[]
for neuron_weight ,neuron_bias in zip(weight,bias):
    neuron_output=0
    for neuron_input,weight in zip(sample,neuron_weight):
        neuron_output+=neuron_input*weight
    neuron_output+=neuron_bias
    layer_outputs.append(neuron_output)
print(layer_outputs)
```
