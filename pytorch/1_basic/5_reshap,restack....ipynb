# reshaping ,stacking,squeezing ,unsqueezing 
1. reshaping- reshaping a input tensor into defined shape
2. view -- return a view of an array to a certain shape but in the shere same  memory  
3. stacking -- combine multiple tensor on top of each other (vstack) or horizontally side by side (h stack)
4. squeezing- removes all 1 dimension form the tensor 
5. unsqueezing --add a 1 dimension to a target tensor
6. permute -- return a view of the input with dimensions permuted(swapped )in certain way 
* all of the thing to make the tensor in certain shape or dimension

```python
import torch
```

```python
x=torch.arange(1.,10.)
```

```python
# reshape
x_reshaped=x.reshape(9,1)
x_reshaped,x.shape
```

```python
# change the view 
z=x.view(1,9)
z.shape
```

```python
z[:,0]=5
x,z
```

```python
x_stacked=torch.stack([x,x,x,x,x,x,x],dim=0)
x_stacked
```

```python
# squeezing is just the remove the extra unnessary dimension from the tensor
x_squeezed=x_reshaped.squeeze()
x_squeezed.shape,x_reshaped.shape
```

```python
x_original=torch.rand([255,255,3])
x_permutted=x_original.permute(2,1,0)
x_original.shape,x_permutted.shape
```

```python
x_unsqueezed=x_original.unsqueeze(0)
x_unsqueezed.shape,x_original.shape
```
