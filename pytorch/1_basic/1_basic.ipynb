## random tensor

Random tensor is important because the way many nn learn is that they starts at random number and then it adjust it weight and bias to better represent data

```python
random_tensor=torch.rand(1,2,3)
random_tensor
```

```python
random_tensor.ndim
```

```python
random_tensor_image=torch.rand(size=(3,255,255)) #height,width,color channel (rgb)
random_tensor_image.shape,random_tensor_image.ndim
```

```python
random_tensor_image
```
